EDA

```python
#Number of Type of Machine failure
ax=sns.countplot(x="Type",data=df)
ax.bar_label(ax.containers[0])
plt.title('Type of Machine Failures')
plt.ylabel("Count of Failures")
plt.show()
```

```python
plt.title("Number of Machines Failed")
ax=sns.countplot(x="Machine failure", data=df)
ax.bar_label(ax.containers[0])
plt.show()
```

```python
sns.histplot(df["Torque [Nm]"],bins=10,kde=True)
plt.show()
```

Outlier Detection

```python
sensor_cols=['Air temperature [K]','Process temperature [K]', 'Rotational speed [rpm]', 'Torque [Nm]','Tool wear [min]']
for col in sensor_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

+ From the outlier detection we undertand that most of the errors are due to rotational speed or torque being drastically changed as they do not lie in the quartile range
